```python
import csv

input_file = "ElectricCar.csv"
output_file = "ElectricCar_clean.csv"

wanted_fields = [
    "Model",
    "VIN (1-10)",
    "County",
    "City",
    "2020 GEOID",
    "State",
    "Vehicle Location"
]

with open(input_file, "r", newline="", encoding="utf-8") as infile:
    reader = csv.DictReader(infile)

    with open(output_file, "w", newline="", encoding="utf-8") as outfile:
        writer = csv.DictWriter(outfile, fieldnames=wanted_fields)
        writer.writeheader()

        for row in reader:

            # Skip rows with a blank GEOID
            if row["2020 GEOID"] == "":
                continue

            clean_row = {}

            for field in wanted_fields:
                clean_row[field] = row[field]

            writer.writerow(clean_row)
```
