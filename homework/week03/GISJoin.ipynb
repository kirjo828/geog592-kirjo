```python
# Import CSV
import csv

filename = "tabela1613.csv"

with open(filename, "r", encoding="utf-8-sig") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

print("Records found:", len(rows))
print(reader.fieldnames)

# Filter for para
para_rows = []

for row in rows:
    if row["Nível"] == "MU" and row["Município"].endswith("(PA)"):
        para_rows.append(row)

print("Pará municipalities:", len(para_rows))

# Keep only the columns we want
selected_columns = []

for row in para_rows:
    selected_columns.append({
        "cod": row["Cód."],
        "acai": row["Açaí"]
    })

for row in selected_columns[:5]:
    print(row)

# create an empty list
nv = []

# column that we want to check
column = "acai"

# check each row
for row in selected_columns:
    try:
        float(row[column])
    except ValueError:
        nv.append(row[column])

print(nv)

# Look for missing values in the Açaí column
missing_count = 0

for row in selected_columns:
    if row["acai"] == "":
        missing_count += 1

print("Missing Açaí values:", missing_count)

# Look for missing values in the cod column
missing_code_count = 0

for row in selected_columns:
    if row["cod"] == "":
        missing_code_count += 1

print("Missing municipality codes:", missing_code_count)

# Check code type
print(type(selected_columns[0]["cod"]))
print(type(selected_columns[0]["acai"]))

# Clean Data
for row in selected_columns:
    if row["acai"] == "-":
        row["acai"] = "0"

    elif row["acai"] == "...":
        row["acai"] = ""

# Count missing values after cleaning
missing_count = 0

for row in selected_columns:
    if row["acai"] == "":
        missing_count += 1

print("Missing values after cleaning:", missing_count)

# Export CSV
fieldnames = selected_columns[0].keys()

with open("para_acai_clean.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(selected_columns)
```
